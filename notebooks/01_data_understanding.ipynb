# Data Understanding : Google Play Store Dataset

## Overview & Dataset Dimensions
* **File Source:** `../data/raw/googleplaystore.csv`
* **Total Rows (Apps):** 10,841
* **Total Columns:** 13 features

---

## Dataset Structure & Features
The dataset contains 13 columns consisting of 12 String based and 1 float based
* **App:** The name of the mobile application.
* **Category:** The primary category the app belongs to. (Top 3 are FAMILY, GAMES and TOOLS)
* **Rating:** The user rating of the application on a scale from 1.0 to 5.0 (with data anomalies showing up to a max of 19.0)
* **Reviews:** The total number of user reviews submitted for the app.
* **Size:** The file size of the application (varying with device or listed in megabytes).
* **Installs:** The number of user downloads/installations.
* **Type:** Indicates whether the app is `Free` or `Paid`. 10039 are free and 800 paid
* **Price:** The monetary cost of the application.
* **Content Rating:** The target audience age group.
* **Genres:** Sub-categories or specific genre classifications.
* **Last Updated:** The date the application was last updated.
* **Current Ver:** The current version of the application.
* **Android Ver:** The minimum Android version required to run the app.

---

## Data Quality & Missing Value Analysis
A thorough check on missing entries and anomalies revealed the following:
* **Missing Values:**
  * `Rating` has 1,474 missing values (~13.60% of the dataset).
  * `Current Ver` has 8 missing values (~0.07%).
  * `Android Ver` has 3 missing values (~0.03%).
  * `Type` and `Content Rating` each have 1 missing value (~0.01%).
* **Duplicates:** The dataset contains **483 duplicate rows** that will need to be cleaned during preprocessing.
* **Data Anomalies:** The statistical summary for `Rating` reveals a data entry error where the `max` value is `19.0` (surpassing the standard 5.0 rating scale limit), requiring data cleaning and outlier handling.
